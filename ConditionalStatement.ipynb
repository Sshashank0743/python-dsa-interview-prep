```python
## Determine if a years is a leap year using nested condition statement

year = int(input("Enter the year"))

print(year)

if (year % 4 == 0):
    if (year % 100 == 0):
        if (year % 400 == 0):
            print(year, " is a leap year")
        else:
            print(year, " is not a leap year")
    else:
        print(year, " is a leap year")
else:
    print(year, " is not a leap year")
```
