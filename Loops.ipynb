```python
## While loop

count = 3

while count<5:
    print(count)
    count += 1
```

```python
## Loop control statement

## Break

for i in range(9):
    if i==7:
        break
    print(i)
```

```python
## Continue

for i in range(15):
    if i%2==0:
        continue
    print(i)
```

```python
## Nested loop statement


for i in range(2):
    for j in range(5):
        print(f"i:{i} and j:{j}")
```

```python
## Calculate the sum of first n natural number

n = 10
sum = 0
count = 1

while count<=n:
    sum = sum+count
    count = count+1

print("The sum of first n natural numbers are: ", sum)
```
