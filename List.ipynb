```python
veggy = ['potato', 'tomato', 'lady-finger', 'chilly', 'ginger']

print(veggy[0])
print(veggy[2])
print(veggy[-1])
print(veggy[2:5])
```

```python
veggy[2] = 'brinjal'
print(veggy)
```

```python
veggy.append('onion')
print(veggy)
```

```python
veggy.insert(2, 'onion')
print(veggy)
```

```python
print(veggy.count("onion"))
```

```python
veggy.remove('onion')
print(veggy)
```

```python
## remove and print the last item

popped_out = veggy.pop()
print(popped_out)
print(veggy)
```

```python
## get the index value

index = veggy.index("chilly")
print(index)
veggy.sort()
print(veggy)
```

```python
veggy.reverse()
print(veggy)
```

```python
numbers = [1,2,5,8,9,6,3,4,7,9]
print(numbers[2:5])
print(numbers[:6])
print(numbers[6:])
print(numbers[:-1])
print(numbers[3:-2])
print(numbers[::3])
print(numbers[::-1])
print(numbers[::-2])
```

```python
for number in numbers:
    print(number)
```

```python
## Now we can find the index value of all numbers by in-built python function

for index, number in enumerate(numbers):
    print(index, number)
```

```python
## List comprehension

lst = []

for a in range(10):
    lst.append(a**2)

print(lst)
```
